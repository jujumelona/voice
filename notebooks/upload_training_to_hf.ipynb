# Upload trained artifacts to Hugging Face

Run the cell below after training is finished. It asks for a **masked Hugging Face write token** and then uploads every detected epoch checkpoint plus the completed outputs. Each epoch folder gets a README stating that training used **20 Japanese songs**.


```python
from getpass import getpass
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / 'scripts' / 'upload_training_to_huggingface.py').exists():
    raise RuntimeError('Run this notebook from the voice repository root.')
sys.path.insert(0, str(repo_root))

from scripts.upload_training_to_huggingface import upload_training_artifacts

HF_TOKEN = getpass('Hugging Face write token: ')
result = upload_training_artifacts(
    token=HF_TOKEN,
    checkpoint_root='checkpoints',
    final_dirs=('outputs',),
    song_count=20,
)
print(f"Done: https://huggingface.co/{result['repo_id']}")
print(f"Uploaded epochs: {[item['epoch'] for item in result['epochs']]}")
```
